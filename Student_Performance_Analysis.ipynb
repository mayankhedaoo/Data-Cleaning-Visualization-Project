# Student Performance Analysis

Data Cleaning & Visualization Project

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files

uploaded = files.upload()

df = pd.read_csv('student-mat.csv', sep=';')
df.head()
```

```
print("Shape:", df.shape)
print(df.info())
print(df.describe())
```

```
print(df.isnull().sum())
print("Duplicates:", df.duplicated().sum())
df = df.drop_duplicates()
```

```
sns.countplot(x='sex', data=df)
plt.title('Gender Distribution')
plt.show()
```

```
plt.hist(df['age'], bins=8)
plt.title('Age Distribution')
plt.show()
```

```
sns.boxplot(x='studytime', y='G3', data=df)
plt.title('Study Time vs Final Grade')
plt.show()
```

```
sns.scatterplot(x='absences', y='G3', data=df)
plt.title('Absences vs Final Grade')
plt.show()
```

```
sns.barplot(x='Medu', y='G3', data=df)
plt.title("Mother Education vs Final Grade")
plt.show()
```

```
numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(10,8))
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

## Key Findings

1. Higher study time generally improves grades.
2. More absences are associated with lower performance.
3. Parents' education positively influences results.
4. G1 and G2 strongly correlate with final grade G3.
